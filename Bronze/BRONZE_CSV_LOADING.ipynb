## Case Demanding:

- Data Loading:
Load the provided three files using either SQL or PySpark.
Name each table with a raw_ prefix to differentiate between original and transformed data.

#### Environment Setup

```python
from pyspark.sql import functions as F
from pyspark.sql.types import *
from delta.tables import DeltaTable
import datetime
```

```python
(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.raw_products")
  .addColumn("ProductID", "STRING")
  .addColumn("ProductDesc", "STRING")
  .addColumn("ProductNumber", "STRING")
  .addColumn("MakeFlag", "STRING")
  .addColumn("Color", "STRING")
  .addColumn("SafetyStockLevel", "STRING")
  .addColumn("ReorderPoint", "STRING")
  .addColumn("StandardCost", "STRING")
  .addColumn("ListPrice", "STRING")
  .addColumn("Size", "STRING")
  .addColumn("SizeUnitMeasureCode", "STRING")
  .addColumn("Weight", "STRING")
  .addColumn("WeightUnitMeasureCode", "STRING")
  .addColumn("ProductCategoryName", "STRING")
  .addColumn("ProductSubCategoryName", "STRING")
  .addColumn("file_modification_time", "TIMESTAMP")
  .addColumn("file_path", "STRING")
  .addColumn("file_name", "STRING")
  .addColumn("file_size", "BIGINT")
  .addColumn("file_loading_time", "TIMESTAMP")
  .execute()
)

(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.raw_sales_order_detail")
  .addColumn("SalesOrderID", "STRING")
  .addColumn("SalesOrderDetailID", "STRING")
  .addColumn("OrderQty", "STRING")
  .addColumn("ProductID", "STRING")
  .addColumn("UnitPrice", "STRING")
  .addColumn("UnitPriceDiscount", "STRING")
  .addColumn("file_modification_time", "TIMESTAMP")
  .addColumn("file_path", "STRING")
  .addColumn("file_name", "STRING")
  .addColumn("file_size", "BIGINT")
  .addColumn("file_loading_time", "TIMESTAMP")
  .execute()
)

(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.raw_sales_order_header")
  .addColumn("SalesOrderID", "STRING")
  .addColumn("OrderDate", "STRING")
  .addColumn("ShipDate", "STRING")
  .addColumn("OnlineOrderFlag", "STRING")
  .addColumn("AccountNumber", "STRING")
  .addColumn("CustomerID", "STRING")
  .addColumn("SalesPersonID", "STRING")
  .addColumn("Freight", "STRING")
  .addColumn("file_modification_time", "TIMESTAMP")
  .addColumn("file_path", "STRING")
  .addColumn("file_name", "STRING")
  .addColumn("file_size", "BIGINT")
  .addColumn("file_loading_time", "TIMESTAMP")
  .execute()
)
```

#### Products CSV Ingestion

```python
try:
  #Defining product csv schema
  product_schema = StructType([
    StructField("ProductID", StringType(), True),
    StructField("ProductDesc", StringType(), True),
    StructField("ProductNumber", StringType(), True),
    StructField("MakeFlag", StringType(), True),
    StructField("Color", StringType(), True),
    StructField("SafetyStockLevel", StringType(), True),
    StructField("ReorderPoint", StringType(), True),
    StructField("StandardCost", StringType(), True),
    StructField("ListPrice", StringType(), True),
    StructField("Size", StringType(), True),
    StructField("SizeUnitMeasureCode", StringType(), True),
    StructField("Weight", StringType(), True),
    StructField("WeightUnitMeasureCode", StringType(), True),
    StructField("ProductCategoryName", StringType(), True),
    StructField("ProductSubCategoryName", StringType(), True)
  ])

  #Reading product csv, with _metadata columns to better log of files
  df_new_products = (
    spark.read.format("csv")
    .option("header", True)
    .schema(product_schema)
    .option("delimiter", ",")
    .load("/Volumes/interviewcaseupstart/sales_db/csv_files/products.csv")
    .withColumn("file_modification_time", F.col("_metadata.file_modification_time"))
    .withColumn("file_path", F.col("_metadata.file_path"))
    .withColumn("file_name", F.col("_metadata.file_name"))
    .withColumn("file_size", F.col("_metadata.file_size"))
    .withColumn("file_loading_time", F.lit(datetime.datetime.now().strftime("%Y-%m-%dT%H:%M:%S")).cast(TimestampType()))
  )

  #Apending raw data to raw_products
  df_new_products.write.mode("append").saveAsTable("interviewcaseupstart.sales_db.raw_products")

except Exception as e:
  dbutils.notebook.exit(e)
```
