```python
from delta.tables import DeltaTable, IdentityGenerator
import os
```

```python
%sql
CREATE CATALOG IF NOT EXISTS interviewcaseupstart;
CREATE SCHEMA IF NOT EXISTS interviewcaseupstart.sales_db;
CREATE VOLUME IF NOT EXISTS interviewcaseupstart.sales_db.csv_files;
```

```python
# Get the absolute path of a file relative to the current working directory
relative_path_products = "../InterviewCaseUpstart/csv_files/products.csv"
relative_path_order_detail = "../InterviewCaseUpstart/csv_files/sales_order_detail.csv"
relative_path_order_header = "../InterviewCaseUpstart/csv_files/sales_order_header.csv"
absolute_path_products = os.path.abspath(relative_path_products)
absolute_path_order_detail = os.path.abspath(relative_path_order_detail)
absolute_path_order_header = os.path.abspath(relative_path_order_header)

dbutils.fs.cp(absolute_path_products, "/Volumes/interviewcaseupstart/sales_db/csv_files/products.csv")
dbutils.fs.cp(absolute_path_order_detail, "/Volumes/interviewcaseupstart/sales_db/csv_files/sales_order_detail.csv")
dbutils.fs.cp(absolute_path_order_header, "/Volumes/interviewcaseupstart/sales_db/csv_files/sales_order_header.csv")
```

```python
(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.raw_products")
  .addColumn("ProductID", "STRING")
  .addColumn("ProductDesc", "STRING")
  .addColumn("ProductNumber", "STRING")
  .addColumn("MakeFlag", "STRING")
  .addColumn("Color", "STRING")
  .addColumn("SafetyStockLevel", "STRING")
  .addColumn("ReorderPoint", "STRING")
  .addColumn("StandardCost", "STRING")
  .addColumn("ListPrice", "STRING")
  .addColumn("Size", "STRING")
  .addColumn("SizeUnitMeasureCode", "STRING")
  .addColumn("Weight", "STRING")
  .addColumn("WeightUnitMeasureCode", "STRING")
  .addColumn("ProductCategoryName", "STRING")
  .addColumn("ProductSubCategoryName", "STRING")
  .addColumn("file_modification_time", "TIMESTAMP")
  .addColumn("file_path", "STRING")
  .addColumn("file_name", "STRING")
  .addColumn("file_size", "BIGINT")
  .addColumn("file_loading_time", "TIMESTAMP")
  .addColumn("file_loaded_store", "BOOLEAN")
  .partitionedBy("file_loaded_store")
  .execute()
)

(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.raw_sales_order_detail")
  .addColumn("SalesOrderID", "STRING")
  .addColumn("SalesOrderDetailID", "STRING")
  .addColumn("OrderQty", "STRING")
  .addColumn("ProductID", "STRING")
  .addColumn("UnitPrice", "STRING")
  .addColumn("UnitPriceDiscount", "STRING")
  .addColumn("file_modification_time", "TIMESTAMP")
  .addColumn("file_path", "STRING")
  .addColumn("file_name", "STRING")
  .addColumn("file_size", "BIGINT")
  .addColumn("file_loading_time", "TIMESTAMP")
  .addColumn("file_loaded_store", "BOOLEAN")
  .partitionedBy("file_loaded_store")
  .execute()
)

(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.raw_sales_order_header")
  .addColumn("SalesOrderID", "STRING")
  .addColumn("OrderDate", "STRING")
  .addColumn("ShipDate", "STRING")
  .addColumn("OnlineOrderFlag", "STRING")
  .addColumn("AccountNumber", "STRING")
  .addColumn("CustomerID", "STRING")
  .addColumn("SalesPersonID", "STRING")
  .addColumn("Freight", "STRING")
  .addColumn("file_modification_time", "TIMESTAMP")
  .addColumn("file_path", "STRING")
  .addColumn("file_name", "STRING")
  .addColumn("file_size", "BIGINT")
  .addColumn("file_loading_time", "TIMESTAMP")
  .addColumn("file_loaded_store", "BOOLEAN")
  .partitionedBy("file_loaded_store")
  .execute()
)

(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.store_products")
  .addColumn("ProductID", "BIGINT", generatedByDefaultAs=IdentityGenerator(),nullable=False)
  .addColumn("ProductDesc", "STRING")
  .addColumn("ProductNumber", "STRING")
  .addColumn("MakeFlag", "BOOLEAN")
  .addColumn("Color", "STRING")
  .addColumn("SafetyStockLevel", "INTEGER")
  .addColumn("ReorderPoint", "INTEGER")
  .addColumn("StandardCost", "DECIMAL(10,4)")
  .addColumn("ListPrice", "DECIMAL(10,3)")
  .addColumn("Size", "STRING")
  .addColumn("SizeUnitMeasureCode", "STRING")
  .addColumn("Weight", "DECIMAL(10,2)")
  .addColumn("WeightUnitMeasureCode", "STRING")
  .addColumn("ProductCategoryName", "STRING")
  .addColumn("ProductSubCategoryName", "STRING")
  .addColumn("LastModificationTime", "TIMESTAMP")
  .addColumn("ProductPublished", "BOOLEAN")
  .partitionedBy("ProductPublished")
  .execute()
)

(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.store_sales_order_detail")
  .addColumn("SalesOrderID", "BIGINT",generatedByDefaultAs=IdentityGenerator(), nullable=False)
  .addColumn("SalesOrderDetailID", "BIGINT",generatedByDefaultAs=IdentityGenerator(), nullable=False)
  .addColumn("OrderQty", "INTEGER")
  .addColumn("ProductID", "BIGINT",nullable=False)
  .addColumn("UnitPrice", "DECIMAL(10,4)")
  .addColumn("UnitPriceDiscount", "DECIMAL(10,4)")
  .addColumn("LastModificationTime", "TIMESTAMP")
  .addColumn("ProductPublished", "BOOLEAN")
  .partitionedBy("ProductPublished")
  .execute()
)

(
  DeltaTable.createIfNotExists(spark)
  .tableName("interviewcaseupstart.sales_db.store_sales_order_header")
  .addColumn("SalesOrderID", "BIGINT", nullable=False)
  .addColumn("OrderDate", "DATE")
  .addColumn("ShipDate", "DATE")
  .addColumn("OnlineOrderFlag", "BOOLEAN")
  .addColumn("AccountNumber", "STRING")
  .addColumn("CustomerID", "BIGINT", nullable=False)
  .addColumn("SalesPersonID", "BIGINT")
  .addColumn("Freight", "DECIMAL(10,4)")
  .addColumn("LastModificationTime", "TIMESTAMP")
  .addColumn("ProductPublished", "BOOLEAN")
  .partitionedBy("ProductPublished")
  .execute()
)
```

```python
%sql
ALTER TABLE interviewcaseupstart.sales_db.store_products ADD CONSTRAINT pk_Product PRIMARY KEY(ProductID);

ALTER TABLE interviewcaseupstart.sales_db.store_sales_order_detail ADD CONSTRAINT pk_Orders PRIMARY KEY(SalesOrderDetailID);

ALTER TABLE interviewcaseupstart.sales_db.store_sales_order_detail ADD CONSTRAINT fk_Product
FOREIGN KEY (ProductID) REFERENCES interviewcaseupstart.sales_db.store_products(ProductID);
```
